```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler
import pandas as pd
```

```python
s=pd.Series([1000,2000,3000,5000,9000])
s
```

# NORMALIZATION

```python
s.min()
```

```python
s.max()
```

```python
s_norm=(s-s.min())/(s.max()-s.min())
print(s_norm)
```

```python
s_norm=MinMaxScaler().fit_transform(s.values.reshape(-1,1))
print(s_norm)
```

```python
df=pd.DataFrame({
    "temp":[20,25,30,35,40],
    "humidity":[30,45,60,70,90]
})
df
```

```python
df_norm=MinMaxScaler().fit_transform(df)
df_norm_df=pd.DataFrame(df_norm,columns=df.columns)
print(df_norm_df)
```

# STANDARDIZATION

```python
s=pd.Series([1000,2000,3000,5000,9000])
s
```

```python
s.mean()
```

```python
s.std()
```

```python
Z_score=(s-s.mean())/s.std()
print(Z_score)
```

```python
Z_score=StandardScaler().fit_transform(s.values.reshape(-1,1))
print(Z_score)
```

```python
df=pd.DataFrame({
    "temp":[20,25,30,35,40],
    "humidity":[30,45,60,70,90]
})
df
```

```python
Z_score=StandardScaler().fit_transform(df)
Z_score_df=pd.DataFrame(Z_score,columns=df.columns)
print(Z_score_df)
```

# OUTLIERS

# 1.IQR METHOD

```python
q1=data.quantile(0.25)
q3=data.quantile(0.75)
IQR=q3-q1
LB=q1-1.5*IQR
UB=q3+1.5*IQR
outliers=data[(data<(LB))|(data>(UB))]

q1
q3
outliers
```
